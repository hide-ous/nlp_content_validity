```python
train = pd.read_excel('../../data/external/pillet_et_al/RATER-C/data/processed/train_val_test.xlsx', sheet_name='train')
test = pd.read_excel('../../data/external/pillet_et_al/RATER-C/data/processed/train_val_test.xlsx', sheet_name='test')
val = pd.read_excel('../../data/external/pillet_et_al/RATER-C/data/processed/train_val_test.xlsx', sheet_name='val')
```
